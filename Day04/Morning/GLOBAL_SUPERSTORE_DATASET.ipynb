```python
# ============================================================
# ETL - GLOBAL SUPERSTORE DATASET
# Step 01: EXTRACT
# ============================================================

import pandas as pd
import numpy as np

# Load the dataset
file_path = "Global_Superstore2.csv"

df = pd.read_csv(file_path, encoding="latin1")

print("Dataset extracted successfully!")

# Display basic information
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display first 5 records
display(df.head())
```

```python
# ============================================================
# STEP 02: DATA PROFILING
# ============================================================

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# ============================================================
# STEP 03: DATA QUALITY CHECK
# ============================================================

# Check unique values in categorical columns
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{'='*60}")
    print(f"Column: {column}")
    print("Unique values:", df[column].nunique())
    print(df[column].dropna().unique()[:20])
```

```python
# Check numeric columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_columns.tolist())

display(df[numeric_columns].describe())
```

```python
# Check columns containing "date"
date_columns = [col for col in df.columns if "date" in col.lower()]

print("Possible date columns:")
print(date_columns)
```

```python
for column in date_columns:
    print(f"\n{column}")
    print(df[column].head())
```

```python
# ============================================================
# STEP 03: DATA TRANSFORMATION
# ============================================================

# Create a copy of the raw dataset
df_clean = df.copy()

print("Original rows:", len(df_clean))
```

```python
# Remove duplicate rows
duplicates = df_clean.duplicated().sum()

print("Duplicate rows found:", duplicates)

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
# Remove leading/trailing spaces from text columns

text_columns = df_clean.select_dtypes(include="object").columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Text values standardized successfully.")
```

```python
# Identify date columns
date_columns = [col for col in df_clean.columns if "date" in col.lower()]

print("Date columns:", date_columns)
```

```python
# ============================================================
# TRANSFORMATION: DATE STANDARDIZATION
# ============================================================

date_columns = ["Order Date", "Ship Date"]

for column in date_columns:
    df_clean[column] = pd.to_datetime(
        df_clean[column],
        dayfirst=True,
        errors="coerce"
    )

print("Date columns converted successfully.")

for column in date_columns:
    print(column, ":", df_clean[column].dtype)
```

```python
# Check the converted dates
display(df_clean[["Order Date", "Ship Date"]].head(10))
```

```python
# ============================================================
# CHECK MISSING VALUES AFTER TRANSFORMATION
# ============================================================

missing_values = df_clean.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])
```

```python
# ============================================================
# TRANSFORMATION: HANDLE MISSING POSTAL CODES
# ============================================================

# Convert Postal Code to string
df_clean["Postal Code"] = df_clean["Postal Code"].astype("Int64").astype("string")

# Replace missing postal codes
df_clean["Postal Code"] = df_clean["Postal Code"].fillna("Unknown")

print("Missing Postal Codes after transformation:")
print(df_clean["Postal Code"].isnull().sum())

print("\nPostal Code sample:")
print(df_clean["Postal Code"].head(10))
```

```python
# ============================================================
# TRANSFORMATION: CHECK INVALID NUMERIC VALUES
# ============================================================

numeric_columns = df_clean.select_dtypes(include=np.number).columns

for column in numeric_columns:
    negative_count = (df_clean[column] < 0).sum()
    zero_count = (df_clean[column] == 0).sum()

    print(f"{column}")
    print(f"  Negative values: {negative_count}")
    print(f"  Zero values: {zero_count}")
```

```python
# ============================================================
# CHECK SHIPPING COST VALUES
# ============================================================

print("Records with zero Shipping Cost:")
display(df_clean[df_clean["Shipping Cost"] == 0])
```

```python
# Check negative profit records
print("Number of negative profit records:")
print((df_clean["Profit"] < 0).sum())

print("\nSample negative profit records:")
display(df_clean[df_clean["Profit"] < 0][
    ["Sales", "Quantity", "Discount", "Shipping Cost", "Profit"]
].head(10))
```

```python
# ============================================================
# TRANSFORMATION: CREATE DATE DIMENSIONS
# ============================================================

# Create Year
df_clean["Order Year"] = df_clean["Order Date"].dt.year

# Create Month
df_clean["Order Month"] = df_clean["Order Date"].dt.month

# Create Month Name
df_clean["Order Month Name"] = df_clean["Order Date"].dt.month_name()

# Create Quarter
df_clean["Order Quarter"] = df_clean["Order Date"].dt.quarter

# Create Year-Month
df_clean["Order Year-Month"] = df_clean["Order Date"].dt.to_period("M").astype(str)

print("Date-based columns created successfully!")

display(df_clean[
    ["Order Date", "Order Year", "Order Month",
     "Order Month Name", "Order Quarter", "Order Year-Month"]
].head(10))
```

```python
# ============================================================
# TRANSFORMATION: CALCULATE PROFIT MARGIN
# ============================================================

df_clean["Profit Margin"] = (
    df_clean["Profit"] / df_clean["Sales"]
) * 100

print("Profit Margin calculated successfully!")

display(df_clean[
    ["Sales", "Profit", "Profit Margin"]
].head(10))
```

```python
# ============================================================
# FINAL TRANSFORMATION CHECK
# ============================================================

print("Original dataset shape:", df.shape)
print("Transformed dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nFinal columns:")
print(df_clean.columns.tolist())
```

```python
# ============================================================
# STEP 04: LOAD
# ============================================================

output_file = "Global_Superstore_Cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Data loaded successfully!")
print("Output file:", output_file)
```
